```python
import os
import csv

def create_csv(face_dir, hand_dir, body_dir, output_csv_path):
    face_files = sorted([f for f in os.listdir(face_dir) if f.endswith('.npy')])

    rows = []
    for face_file in face_files:
        # Derive base name (e.g., "video123" from "video123_face.npy")
        base_name = face_file.replace('_face.npy', '')

        # Construct file paths
        face_path = os.path.join(face_dir, f"{base_name}_face.npy")
        left_hand_path = os.path.join(hand_dir, f"{base_name}_hand1.npy")
        right_hand_path = os.path.join(hand_dir, f"{base_name}_hand2.npy")
        body_path = os.path.join(body_dir, f"{base_name}_pose.npy")

        # Check if all files exist
        if all(os.path.exists(p) for p in [face_path, left_hand_path, right_hand_path, body_path]):
            row = f"{face_path}\t{left_hand_path}\t{right_hand_path}\t{body_path}"
            rows.append([row])
        else:
            print(f"Missing one or more files for base: {base_name}")

    # Write to CSV
    with open(output_csv_path, 'w', newline='') as f:
        writer = csv.writer(f)
        writer.writerows(rows)

    print(f"CSV saved to: {output_csv_path}")

# Example usage
create_csv(
    face_dir='face_features_data',
    hand_dir='hands_features_data',
    body_dir='body_features_data',
    output_csv_path='poses.csv'
)
```
